# satviz launcher
Two ways to get a live globe out of the notebook — neither renders inline.

```
pip install -r requirements.txt
```

## 1. Web viewer (CesiumJS) — opens a link in your browser
Best imagery, day/night lighting, built-in time control. Works locally and on Colab.

````
from satviz_web import launch
srv = launch(groups=["stations", "gps-ops", "starlink"])   # opens http://localhost:8765/
````

````
srv.stop()
````

## 2. Desktop window (PyVista/VTK + Qt) — local machine only

````
from satviz_gui import launch as launch_gui
app = launch_gui(groups=["stations", "gps-ops", "starlink"], trail_minutes=45)
````

````
app.scene.focus(app.scene.find("ISS")[0])       # fly to the ISS
app.scene.speed = 300                              # 5 min of orbit per second
````

````
app.close()
````

## Raw positions for analysis

````
from datetime import datetime, timezone
from satviz import Catalog
cat = Catalog(); cat.fetch(["stations"])
cat.propagate([datetime.now(timezone.utc)])[:, 0, :]   # ECI km, shape (n, 3)
````
