# satviz — launch the standalone viewer
Opens a separate window (PyVista/VTK + Qt). The notebook stays usable while it runs.

```
pip install sgp4 numpy requests matplotlib pyvista pyvistaqt PyQt5
```

````
from satviz_gui import launch
app = launch(groups=["stations", "gps-ops", "starlink"], trail_minutes=45)
````

**Mouse:** left-drag orbit · scroll zoom · middle-drag (or shift+drag) pan · click a satellite to label it.

**Panel:** pause/play, speed (×1 … ×3600), time-offset scrub, trail length, group visibility, ECI vs Earth-fixed frame, find/focus, refresh TLEs.

You can also drive it from here:

````
app.scene.focus(app.scene.find("ISS")[0])       # fly to the ISS
````

````
app.scene.speed = 300                              # 5 min of orbit per second
````

````
app.scene.set_frame(earth_fixed=True)               # Earth-fixed view: trails become ground tracks
````

````
pos = app.catalog.propagate([app.scene.now()])[:, 0, :]   # raw ECI positions (km) for analysis
pos.shape
````

````
app.close()
````
