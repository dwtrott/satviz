# satviz on Google Colab
Runs the web viewer inside Colab and gives you a link to open in a new tab. Nothing renders inline.

```
!git clone -q https://github.com/dwtrott/satviz.git
%cd satviz
!pip install -q requests
```

```
from satviz_web import launch
srv = launch(groups=["stations", "gps-ops", "starlink"])   # prints a clickable link
```

The link is a Colab-proxied URL; it works while this runtime is alive and you're signed in to the same Google account.

In the page: drag to orbit, scroll to zoom, right-drag to tilt/pan, click a satellite for details, use the clock widget (bottom-left) to speed up time and the timeline to scrub.
